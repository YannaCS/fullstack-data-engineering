Python type hints are a way to annotate what types your variables, function inputs, and outputs are supposed to be, without changing how Python runs at runtime.

They’re mainly for:
- Readability
- Static type checking
- Better IDE autocomplete
- Fewer bugs in large codebases

## Common places to use type hints

### 1️⃣ Function parameters & return types


````python
def my_add(a: int, b: int) -> int:
    return a + b
````

Python does not enforce this at runtime:

````python
my_add("1", "2")  # runs, but type checkers will complain
````

### 2️⃣ Variables
```python
age: int = 25
price: float = 19.99
is_active: bool = True
```
Useful when the value isn’t obvious:
```python
users: list[str] = []
```

## Typed dictionaries (TypedDict)

````python
from typing import TypedDict

class User(TypedDict):
    id: int
    name: str
    active: bool
````

## Callable

````python
from typing import Callable

def run(task: Callable[[int, int], int]) -> int:
    return task(1, 2)
````

## Generics (TypeVar)

````python
from typing import TypeVar, List

T = TypeVar("T")

def first(items: List[T]) -> T:
    return items[0]
````

## Dataclasses with type hints

````python
from dataclasses import dataclass

@dataclass
class User:
    id: int
    name: str
    active: bool = True
````

## Annotated

`typing.Annotated`
- `Annotated[T, x, y, …]` means: “This value is of type T, plus there is metadata x, y, … associated with it.”  ￼
- The metadata is arbitrary and can be anything (objects, classes, values).  ￼
- This metadata is optional for tools that don’t care about it:
  - Static type checkers (like mypy) usually just ignore it and treat Annotated[T, …] as T.  ￼
  - Runtime code (or libraries) can inspect it if they want.
  - useful for fastapi

````python
from typing import Annotated, get_type_hints

class Range:
    def __init__(self, lo: int, hi: int):
        self.lo = lo
        self.hi = hi

# Annotate an int with a range
Age = Annotated[int, Range(0, 120)]

def set_age(age: Age):
    return age

# Get type hints including metadata
hints = get_type_hints(set_age, include_extras=True)
print(hints)  
# → {'age': typing.Annotated[int, Range(0, 120)]}

# Access metadata
annot = hints['age']
origin = getattr(annot, "__origin__", None)         # int
metadata = getattr(annot, "__metadata__", None)     # (Range(0, 120),)
print(origin)
print(metadata)
````

Annotated does not enforce anything by itself.   
**The metadata you attach is not validated by Python**, it’s up to whatever tool or library reads it

````python
my_age = set_age(-1)
my_age
````

# Importance
for interviews, recommend not to use type:
- not all python versions support
- a lot online IDE not support
- cost more time

Type hints ≠ runtime type checking

Important mindset shift:
- ❌ Type hints do not stop wrong types at runtime
- ✅ They help before code runs

To enforce types, you need:
- Static checker: mypy, pyright
- Runtime validation: pydantic, attrs
