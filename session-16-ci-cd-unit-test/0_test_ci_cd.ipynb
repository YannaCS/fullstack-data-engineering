# Part I: Testing Fundamentals

## 1. Software Testing Overview

### What is Software Testing?

**Software testing** is the process of evaluating and verifying that a software application or system:
- Meets specified requirements
- Works as expected
- Can be implemented with the same characteristics

**Why Testing Matters:**
- **Early bug detection**: Find issues before production
- **Cost reduction**: Cheaper to fix bugs early
- **Quality assurance**: Ensures software meets standards
- **User satisfaction**: Better user experience
- **Risk mitigation**: Reduces production failures

### Testing vs Debugging

| Testing | Debugging |
|  -- |   --|
| Identifies that bugs exist | Finds and fixes the bugs |
| Done by testers | Done by developers |
| Planned activity | Reactive activity |
| Finds unknown issues | Fixes known issues |

### QA vs QC

- **Quality Assurance (QA)**: Process-oriented, **prevents** defects
  - Reviews, audits, processes
  - "Are we building the product right?"
  
- **Quality Control (QC)**: Product-oriented, **detects** defects
  - Testing, inspections
  - "Are we building the right product?"

 


## 2. Testing Techniques

Testing techniques determine **how** you design and approach your tests.

<img src='./images/black_white_gray_box_testing.webp' width=600>

### Black Box Testing

**Definition**: Testing from the **user's perspective** without knowledge of internal code structure.

**Philosophy**: "I don't care HOW it works, I care IF it works"

**Focus Areas**:
- Input-output relationships
- Functionality and requirements
- User experience
- Specification compliance

**Real-World Analogy**: 
Testing a car by driving it. You don't need to know how the engine works internally; you just check if pressing the gas pedal makes it go faster.

**When to Use**:
- ✅ User acceptance testing (UAT)
- ✅ System testing
- ✅ Functional testing
- ✅ Testing third-party APIs/libraries

**Advantages**:
- No programming knowledge required
- Unbiased (tester isn't influenced by code)
- Tests from user perspective
- Can start early (only needs requirements)

**Disadvantages**:
- Cannot test internal logic paths
- May miss edge cases in code
- Limited code coverage visibility

**Example**:
```python
    # Testing a login function (Black Box)
    def test_login_valid_credentials():
        # Don't know how authenticate() works internally
        result = authenticate("valid_user", "valid_password")
        assert result == True

    def test_login_invalid_credentials():
        result = authenticate("invalid_user", "wrong_password")
        assert result == False

    def test_login_empty_credentials():
        result = authenticate("", "")
        assert result == False
```

 



### White Box Testing

**Definition**: Testing with **full knowledge** of internal code structure, logic, and implementation.

**Philosophy**: "I need to verify every path and decision in the code"

**Focus Areas**:
- Code paths and branches
- Conditional logic (if/else)
- Loop execution
- Internal data structures
- Code coverage metrics

**Real-World Analogy**: 
Testing a car by opening the hood, checking each component, verifying the engine's combustion cycle, testing each sensor independently.

**When to Use**:
- ✅ **Unit testing**
- ✅ Security testing (finding vulnerabilities)
- ✅ Code optimization
- ✅ Refactoring validation

**Advantages**:
- Thorough testing of logic
- Find hidden errors
- Optimize code performance
- Measurable (code coverage %)

**Disadvantages**:
- Requires programming knowledge
- Time-consuming
- May miss requirement gaps
- Can't test missing functionality

**Code Coverage Metrics**:
- **Statement Coverage**: Every line executed
- **Branch Coverage**: Every if/else path tested
- **Path Coverage**: Every possible execution path
- **Condition Coverage**: Every boolean sub-expression

**Example**:
```python
# Function with multiple branches (White Box)
def calculate_discount(price, user_type):
    if price > 100:  # Branch 1
        if user_type == "premium":  # Branch 2
            return price * 0.8  # 20% discount
        else:
            return price * 0.9  # 10% discount
    else:  # Branch 3
        return price  # No discount

# White box tests cover ALL branches
def test_premium_high_price():  # Covers Branch 1 + 2
    assert calculate_discount(150, "premium") == 120

def test_regular_high_price():  # Covers Branch 1 + else
    assert calculate_discount(150, "regular") == 135

def test_low_price():  # Covers Branch 3
    assert calculate_discount(50, "premium") == 50
```

 



### Gray Box Testing

**Definition**: Hybrid approach combining black box and white box techniques.

**What it is**: Test with **partial knowledge** of internal workings
- Know database schema but not application logic
- Understand architecture but not implementation details

**When to Use**:
- ✅ Integration testing
- ✅ Security testing
- ✅ API testing with knowledge of backend structure

**Example**: Testing a web application knowing the database schema but not the application code logic.

 



## 3. Testing Levels (classification)

- Testing levels define the **scope** and **granularity** of what you're testing.
- Testing levels describe layers of testing based on system scope.
- what I am testing

```text
┌──────────────────────────────────────────┐
│        E2E Tests (System Level)          │  ← Full user workflows
├──────────────────────────────────────────┤
│    System Tests (Complete Application)   │  ← Entire system
├──────────────────────────────────────────┤
│  Integration Tests (Components Together) │  ← Component interactions
├──────────────────────────────────────────┤
│    Unit Tests (Individual Functions)     │  ← Smallest units
└──────────────────────────────────────────┘
```



### 3.1 Unit Testing
**Scope**: Individual functions, methods, or classes in **isolation**

**Characteristics**:
- Fast execution (milliseconds)
- No external dependencies
- Tests single responsibility
- Highly focused
- Use **mocking** to isolate dependencies
- Focus on **business logic**
- Follow **F.I.R.S.T** Principle
  - Fast: the faster, the better
  - Independent：The lower the dependency, the better
  - Repeatable：Reduce environmental impact
  - Self-Validating：Output Boolean
  - Timely (the concept of TDD) 
    - Timely TDD (Test-Driven Development) is a development process, simply, it is "write tests first and then develop". 
    - In addition to ensuring that the test program is written, writing tests first has another benefit: it helps clarify how the program interface is designed in the early stages of development.

**Example**: Test a single function   
```python
# test_utils.py
def test_add_two_positive(self):
    # Arrange
    # Action
    result = add(1, 2)
    expected = 3
    # Assert
    assert result == expected
```
 

#### Unit Testing Strategy

**What to Test**:
- Business logic calculations
- Data transformations
- Validation functions
- Utility functions
- Error handling

**What NOT to Test** (use integration tests instead):
- Database queries
- External API calls
- File system operations
- Network communication

#### Unit Testing with Mocking

When function has dependencies, **mock them**:

```python
from unittest.mock import Mock, patch

def test_get_user_data():
    # Arrange: Mock the database
    mock_db = Mock()
    mock_db.query.return_value = {"id": 1, "name": "Alice"}
    
    service = UserService(db=mock_db)
    
    # Act
    result = service.get_user(1)
    
    # Assert
    assert result["name"] == "Alice"
    mock_db.query.assert_called_once_with("SELECT * FROM users WHERE id=1")
```

### 3.2 Integration Testing
**Scope**: Testing **interactions between components**   
- Service + Database
- Service + External API
- Multiple services communicating
- Component integration


**Characteristics**:
- **Slower** than unit tests
- Tests component interfaces
- Verify **data flow** between layers
- May use **real dependencies**, or
- **Mock external dependencies** (APIs, databases, internet, servers)
- Use **TestContainers** for realistic environments

**Example**: Test database connection + data access layer

 ```python
# test_services.py
def test_get_weather(self, service, mock_db, mock_api_client):
    # Arrange
    mock_api_client.get_current_weather.return_value = {...}
    
    # Action
    result = service.get_weather('Tokyo')
    
    # Assert
    assert result['city'] == 'Tokyo'
    mock_api_client.get_current_weather.assert_called_once_with('Tokyo')
    mock_db.add.assert_called_once()
```


#### Integration Testing Strategy
**When to Use**:   
- Testing database operations
- Testing API endpoints
- Testing message queue producers/consumers
- Testing service-to-service communication

#### TestContainers: Real Dependencies in Tests

- Testcontainers is a testing library that let us run **real infrastructure components** (Docker containers) during automated tests.
- Instead of mocking databases, Kafka, Redis, etc., we spin up the real thing, run tests, then tear it down automatically.

**Benefits**:
- ✅ Tests against real behavior
- ✅ Catches database-specific issues
- ✅ Tests real transactions and constraints
- ✅ Automatic cleanup after tests

**TestContainers Examples:**   

```bash
    pip install testcontainers[postgresql] psycopg2-binary
```
Make sure Docker is running.   



*PostgreSQL example (most common)*  
```python
from testcontainers.postgres import PostgresContainer
import psycopg2

def test_postgres_container():
    # Start a real PostgreSQL container
    with PostgresContainer("postgres:15") as postgres:
        conn = psycopg2.connect(
            host=postgres.get_container_host_ip(),
            port=postgres.get_exposed_port(5432),
            user=postgres.USER,
            password=postgres.PASSWORD,
            database=postgres.DBNAME,
        )

        cur = conn.cursor()
        cur.execute("CREATE TABLE users (id SERIAL PRIMARY KEY, name TEXT);")
        cur.execute("INSERT INTO users (name) VALUES ('Alice');")
        conn.commit()

        cur.execute("SELECT name FROM users;")
        result = cur.fetchone()

        assert result[0] == "Alice"

        cur.close()
        conn.close()
```
✅ Spins up real PostgreSQL      
✅ Runs the test      
✅ Container is destroyed automatically      



*Kafka example (Python)*  
```python
from testcontainers.kafka import KafkaContainer
from confluent_kafka import Producer, Consumer
import time

def test_kafka_produce_consume():
    with KafkaContainer("confluentinc/cp-kafka:7.5.0") as kafka:
        bootstrap = kafka.get_bootstrap_server()

        producer = Producer({"bootstrap.servers": bootstrap})
        producer.produce("test-topic", value=b"hello")
        producer.flush()

        consumer = Consumer({
            "bootstrap.servers": bootstrap,
            "group.id": "test-group",
            "auto.offset.reset": "earliest",
        })

        consumer.subscribe(["test-topic"])

        msg = None
        for _ in range(10):
            msg = consumer.poll(1.0)
            if msg:
                break

        assert msg is not None
        assert msg.value() == b"hello"

        consumer.close()
```
This tests:
- real broker
- real offsets
- real consumer group behavior



*pytest-friendly version (recommended)*
```python
import pytest
from testcontainers.postgres import PostgresContainer

@pytest.fixture(scope="session")
def postgres():
    with PostgresContainer("postgres:15") as container:
        yield container

def test_db(postgres):
    assert postgres.get_exposed_port(5432)
```
✔ Container shared across tests      
✔ Much faster for large suites



### 3.3 System Testing
**Scope**: Testing the **complete, integrated system**

**Characteristics**:
- Tests against requirements
- End-to-end workflows
- Black box approach
- Full environment

**Example**: Test entire application deployment

System tests often overlap with E2E tests. If a system test validates a full user or data flow across all components, it’s effectively an E2E test. However, system tests may still mock external systems, while E2E tests usually exercise real end-to-end integrations.


### 3.4 End-to-End Tests
- Test **complete user flows**
- Use real or staging environments
- Verify system behavior **from user perspective** 
- test UI

**Characteristics**:
- Test from user perspective (**black box**)
- Use real or staging environments
- Test UI + Backend + Database
- Slowest and most expensive

**Focus**: Verify system behavior as a whole


**Example for Backend APIs:**
```python
# test_endpoints.py
def test_get_weather_point(self, client, mock_service):
    # Arrange
    mock_service.get_weather.return_value = {...}
    app.dependency_overrides[get_weather_service] = lambda: mock_service
    
    # Action
    response = client.get('/weather/Tokyo')
    
    # Assert
    assert response.status_code == 200
    assert response.json()['city'] == 'Tokyo'
```



#### E2E Testing Tools

**Backend E2E**:
- Postman/Newman
- REST Assured (Java)
- Supertest (Node.js)
- pytest + requests (Python)

**Frontend E2E** (for comparison):
- Selenium: Browser automation
- Cypress: Modern E2E framework
- Playwright: Cross-browser testing


#### E2E Testing Strategy

**Use E2E for**:
- ✅ Critical user journeys
- ✅ Payment flows
- ✅ Registration/login flows
- ✅ Data migration verification
- ✅ Cross-service workflows

**Don't use E2E for**:
- ❌ Testing every edge case
- ❌ Testing business logic (use unit tests)
- ❌ Testing individual functions
- ❌ Running frequently (too slow)


#### E2E Testing Challenges

**Common Issues**:
- ⚠️ Slow execution (minutes)
- ⚠️ Flaky tests (timing issues)
- ⚠️ Difficult debugging
- ⚠️ Expensive to maintain
- ⚠️ Environment dependencies

**Solutions**:
- Run E2E tests only on staging/production-like environments
- Use **retry logic** for flaky tests
- Implement proper **test isolation**
- Use CI/CD to run E2E tests nightly
- Focus on happy path scenarios

### 3.5 Acceptance Testing
**Scope**: Validation against **business requirements**

**Characteristics**:
- User-focused
- Business validation
- Real-world scenarios
- Sign-off criteria

**Example**: UAT (User Acceptance Testing)

Acceptance testing verifies business requirements. It often uses E2E tests for implementation, but acceptance is defined by intent, not by test level.

## 5. Testing Best Practices

### 5.1 FIRST Principles

Guidelines for **good unit tests**:

- **F**ast: Tests run quickly (milliseconds)
  - Enables frequent execution
  - Fast feedback loop
  
- **I**solated/Independent: Tests don't depend on each other
  - Can run in any order
  - One test failure doesn't affect others
  - No shared state
  
- **R**epeatable: Same result every time
  - No randomness
  - No dependency on environment
  - Deterministic outcomes
  
- **S**elf-validating: Pass/fail is automatic (Boolean output)
  - No manual verification needed
  - Clear assertions
  - Automated checks
  
- **T**imely: Written alongside code (TDD approach)
  - Write tests first, then code
  - Ensures testability
  - Clarifies interface design early

**Timely & TDD (Test-Driven Development)**:
1. Write failing test first
2. Write minimum code to pass
3. Refactor while keeping tests green

**Benefits of writing tests first**:
- Ensures test coverage
- Clarifies API design early
- Prevents over-engineering
- Documents expected behavior



### 5.2 AAA Pattern (Arrange-Act-Assert)
A **structure** for writing clear, readable tests, **especially unit tests**.   

**Why AAA?**
- Clear test structure
- Easy to read
- Separates setup from validation
- Makes intent obvious

**Where AAA is used**  
- Unit tests (most common)
- Integration tests (still helpful)
- Testcontainers tests
- Pytest, JUnit, NUnit, etc.

**Structure** for organizing test code:   
```python
def test_calculate_discount():
    # ARRANGE: Set up test data and dependencies
    price = 100
    user_type = "premium"
    
    # ACT: Execute the behavior being tested
    result = calculate_discount(price, user_type)
    
    # ASSERT: Verify the outcome
    assert result == 80
```
**The three steps**:       

1️⃣ Arrange: Set up everything the test needs
- input data
- objects
- mocks / stubs
- environment

2️⃣ Act: Execute one action — the behavior under test.   
3️⃣ Assert: Verify the outcome.    

**Common mistakes**:     

- ❌ Multiple Acts in one test
- ❌ Asserting during Arrange
- ❌ Complex logic inside tests
- ❌ Tests that don’t clearly separate steps

**Alternative Patterns**:
- **Given-When-Then** (BDD style): Same concept, different naming
- **Setup-Exercise-Verify**: Similar to AAA

### 5.4 Test Data Management

**Best Practices**:
- Use fixtures for **reusable test data**
  - A **pytest fixture** provides test dependencies (data, objects, resources) with controlled setup and teardown.
  - Think of it as shared Arrange logic for tests
  - **Fixtures manage test dependencies; tests should focus on behavior**.
- Avoid hardcoded values when possible
- Use factories for complex objects
- Clean up after tests

```python
  @pytest.fixture
  def sample_user():
      return User(name="Alice", email="alice@example.com")

  def test_user_creation(sample_user):
      assert sample_user.name == "Alice"
```
- pytest sees sample_user as a function argument
- it runs the fixture first
- injects its return value into the test

Fixture with teardown (yield):
```python
  @pytest.fixture
  def db_connection():
      conn = connect_to_db()
      yield conn      # test runs here
      conn.close()    # teardown
```
yield splits setup / teardown.

### 5.5 Test Coverage

#### What is Test Coverage?

**Definition**: Percentage of code executed by tests

**Formula**: `(Lines executed / Total lines) × 100`



#### Types of Coverage

**1. Statement Coverage** (Most Common):
- % of lines of code executed

**2. Branch Coverage**:
- % of if/else paths taken

**3. Function Coverage**:
- % of functions called

**4. Path Coverage** (Most Thorough):
- % of all possible paths through code



#### Example: Coverage Calculation

```python
def calculate(x):
    if x > 0:        # Branch A
        if x > 10:   # Branch B
            return x * 2
        return x
    return 0         # Branch C

# Test 1:
calculate(-5)  # Covers: Branch C

# Test 2:
calculate(5)   # Covers: Branch A (not B)

# Test 3:
calculate(15)  # Covers: Branch A + B

# 100% Branch Coverage = All 3 tests needed
```


#### Measuring Coverage in Python

```bash
# Run tests with coverage report
pytest --cov=app --cov-report=term-missing

# Generate HTML coverage report
pytest --cov=app --cov-report=html

# View report in browser
open htmlcov/index.html
```

**Output**:
```text
    Name                Stmts   Miss  Cover   Missing
    -------------------------------------------------
    app/services.py        45      3    93%   23-25
    app/utils.py           12      0   100%
    -------------------------------------------------
    TOTAL                  57      3    95%
```



#### Coverage Goals

| Context | Target Coverage |
|---------|----------------|
| **Unit tests** | 80-90% |
| **Critical paths** | 90-100% |
| **Overall project** | 70%+ |
| **New features** | 80%+ from day one |
| **Legacy code** | Start with 50%, improve gradually |

 

#### Coverage Tools by Language

**Python**:
- coverage.py
- pytest-cov

**Java**:
- JaCoCo
- Cobertura

**JavaScript**:
- Istanbul
- Jest --coverage



#### Coverage vs Quality

⚠️ **Critical**: **100% coverage ≠ 100% correct**

**Bad test with 100% coverage**:
```python
def test_add():
    result = add(2, 2)
    assert result != None  # ❌ Useless assertion!
```

**Good test**:
```python
def test_add():
    assert add(2, 2) == 4  # ✅ Meaningful assertion
    assert add(-1, 1) == 0
    assert add(0, 0) == 0
```



#### What Coverage Shows

**Coverage tells you**:
- ✅ Which code was executed
- ✅ Which lines are untested
- ✅ Where to add more tests

**Coverage does NOT tell you**:
- ❌ If tests are meaningful
- ❌ If edge cases are covered
- ❌ If assertions are correct
- ❌ If code is well-designed



#### Best Practices

**Do**:
- ✅ Track coverage trends over time
- ✅ Enforce minimum coverage in CI (e.g., 70%)
- ✅ Focus on testing critical business logic first
- ✅ Use coverage to find untested code

**Don't**:
- ❌ Aim for 100% coverage everywhere
- ❌ Write tests just to increase coverage
- ❌ Ignore code quality for coverage
- ❌ Test getters/setters just for coverage

## 6. Mocking & Test Doubles

Mocking and Test Doubles are techniques used in testing to replace real dependencies with **controlled substitutes**, so tests are fast, isolated, and deterministic.  

- Test doubles is the umbrella term.
- Mocks are one specific type of test double.

### Why Mocking Exists

**The Dependency Problem**:

```text
    Your Code depends on:
    ├── Database (slow, requires setup)
    ├── External API (unreliable, costs money)
    ├── File System (side effects)
    ├── Email Service (don't want to send real emails)
    └── Time/Date (hard to test specific moments)
```

**Solution**: Replace dependencies with **test doubles**



### Mocking vs Testcontainers

| Mocking| Testcontainers| 
| ---------| ------------| 
| Fake dependencies| Real dependencies| 
| **Unit tests**| **Integration tests**| 
| Fast| Slower| 
| Low confidence| High confidence| 

Use both, at different levels


### Types of Test Doubles

```text
    Test Doubles
    ├── Dummy: Placeholder, never actually used
    ├── Stub: Returns hardcoded data
    ├── Spy: Records how it was called
    ├── Mock: Verifies behavior and interactions
    └── Fake: Simplified working implementation
```


#### 1. Stub (Returns Data)

```python
# Stub - returns predefined data
stub_db = Mock()
stub_db.get_user.return_value = User(name="Alice")

service = UserService(db=stub_db)
user = service.get_user(1)

assert user.name == "Alice"
```

#### 2. Mock (Verifies Calls)

```python
# Mock - verifies behavior
mock_email = Mock()

send_welcome_email(mock_email, "test@example.com")

# Verify the method was called correctly
mock_email.send.assert_called_once_with(
    to="test@example.com",
    subject="Welcome!"
)
```



#### 3. Spy (Records Calls)

```python
# Spy - records all interactions
spy_logger = Mock()

process_order(order_id=123, logger=spy_logger)

# Check what was logged
assert spy_logger.info.call_count == 3
spy_logger.info.assert_any_call("Processing order 123")
```



#### 4. Fake (Simplified Implementation)

```python
# Fake - simple working implementation
```

**Using unittest.mock**:


### When to Mock

**✅ Mock**:
- External API calls
- Database connections
- File system operations
- Time-dependent functions (datetime.now())
- Random number generation
- Network requests
- Expensive computations

**❌ Don't Mock**:
- Your own domain models
- Value objects (DTOs, POJOs)
- Simple utilities
- The code you're actually testing




### Mocking Best Practices

**Do**:
- ✅ Mock at boundaries (external services)
- ✅ Keep mocks simple
- ✅ Verify important interactions
- ✅ Use TestContainers for databases instead of mocking

**Don't**:
- ❌ Mock what you don't own
- ❌ Over-mock (makes tests brittle)
- ❌ Mock everything (defeats testing purpose)
- ❌ Create complex mock hierarchies

### What is CI/CD?

**Continuous Integration (CI)** is the practice of **automatically building and testing code changes** as they're committed to version control.

> CD can mean Continuous Delivery or Continuous Deployment.   

**Continuous Deployment (CD)** is the practice of **automatically deploying** code that passes all tests **to production**.    

**Continuous Delivery (CD)**: Code is always in a deployable state, but **release to production is a manual decision**.   

```text
Traditional Flow:
Code → Manual Build → Manual Test → Manual Deploy (risky, slow)

CI/CD Flow:
Code → Auto Build → Auto Test → Auto Deploy (safe, fast)
```


### Key Benefits
- **Early bug detection**: Automated tests catch issues before they reach production
- **Faster release cycles**: Automation reduces manual deployment time
- **Improved code quality**: Consistent testing and quality checks
- **Reduced risk**: Small, frequent changes are easier to troubleshoot
- **Team collaboration**: Shared standards and automated workflows



### CI/CD Pipeline Stages

1. **Source**: Code committed to version control (**Git**)
2. **Build**: Code compiled and dependencies installed
3. **Test**: Automated tests run (unit, integration, end-to-end)
4. **Deploy**: Code deployed to staging/production environments
5. **Monitor**: Track application performance and errors

# Part IV: CI/CD Implementation

## GitHub Actions

### Overview
- GitHub Actions is a **CI/CD platform** integrated directly into **GitHub repositories**. 
- It **automates workflows** based on repository events.
- No separate CI/CD server needed

<img src='./images/github_action.webp' width=500>

### Key Concepts

#### Workflows
- **YAML** files stored in `.github/workflows/`
- Define **automated processes triggered by events**
- Can run on different operating systems (Ubuntu, Windows, macOS)

#### Events (Triggers)
- `push`: Code pushed to repository
- `pull_request`: PR opened, synchronized, or reopened
- `schedule`: Cron-based scheduling
- `workflow_dispatch`: Manual trigger
- `release`: New release created

#### Jobs
- **Independent tasks** that run in **parallel (by default)**
- Each job runs on **a fresh virtual machine (runner)**
- Can be **configured to run sequentially** using `needs`

#### Steps
- Individual tasks within a job
- Can run **commands** or use **actions**
- Execute **sequentially within a job**

#### Actions
- **Reusable** units of code
- Can be from **GitHub Marketplace or custom**
- Common examples: `actions/checkout@v3`, `actions/setup-python@v4`

<img src='./images/github-actions-flowchart-automation.jpg' width=500>



### Basic Workflow Example

```yaml
name: CI Pipeline

on:
  push:
    branches: [ main, develop ]
  pull_request:
    branches: [ main ]

jobs:
  test:
    runs-on: ubuntu-latest
    
    steps:
    - name: Checkout code
      uses: actions/checkout@v3
    
    - name: Set up Python
      uses: actions/setup-python@v4
      with:
        python-version: '3.11'
    
    - name: Install dependencies
      run: |
        python -m pip install --upgrade pip
        pip install -r requirements.txt
    
    - name: Run tests
      run: |
        pytest --cov=app --cov-report=term-missing
    
    - name: Upload coverage
      uses: codecov/codecov-action@v3
      if: success()
```



### Advanced Workflow Features

#### Matrix Strategy
Run tests across multiple versions/configurations:

```yaml
jobs:
  test:         # Defines a job named test
    runs-on: ubuntu-latest  # Each job run uses a fresh Ubuntu VM.
    strategy:   # Controls how the job is executed.
      matrix:   # Defines a set of variables and their possible values
        # GitHub will create separate jobs run in parallel by default
        python-version: ['3.9', '3.10', '3.11'] # version testing
        os: [ubuntu-latest, macos-latest]       # os
        requests-version: ['2.28.2', 'latest']  # dependency variations
        
    steps:  # Steps are executed inside each matrix job
    - uses: actions/setup-python@v4 # an official GitHub Action that installs Python
      with:
        python-version: ${{ matrix.python-version }}
    - name: Install dependencies
      run: |
        pip install -U pip
        if [ "${{ matrix.requests-version }}" = "latest" ]; then
          pip install requests
        else
          pip install requests==${{ matrix.requests-version }}
        fi
```
- A matrix strategy allows us to test the same workflow across **multiple environments or configurations** without duplicating job definitions. 
- It’s commonly used for 
  - cross-version testing, 
  - OS compatibility,
  - dependency variations.

#### Environment Variables and Secrets

how GitHub Actions environment variables and secrets are used to securely pass configuration into a job:  

```yaml
jobs:
  deploy:
    runs-on: ubuntu-latest
    env:  # Defines environment variables available to all steps in this job
      DATABASE_URL: ${{ secrets.DATABASE_URL }}
      API_KEY: ${{ secrets.OPENWEATHERMAP_API_KEY }}
    
    steps:
    - name: Deploy application
      run: |  # Runs a multiline shell script.
        echo "Deploying with API key" 
```

- Secrets
  - sensitive values stored securely in GitHub 
  - secrets.*
  - Repository → Settings → Secrets and variables → Actions
- Environment variables (env)
  - how those values are made available to the job and steps
  - DATABASE_URL and API_KEY become normal environment variables
  - They are accessible as $DATABASE_URL and $API_KEY in shell commands
- Steps 
  - can read those variables without hardcoding credentials

| Use case| Use| 
|   -|  -| 
| Sensitive data| secrets.*| 
| Non-sensitive config| env| 
| Per-step overrides| Step-level env| 
| Shared constants| Workflow-level env| 

GitHub Actions secrets allow use to securely inject sensitive configuration into workflows via environment variables. Secrets are encrypted at rest, masked in logs, and scoped appropriately, enabling safe deployment without hardcoding credentials.


#### Conditional Execution

running steps only when certain conditions are true

```yaml
steps:
# The workflow was triggered by a push (or similar) to main
- name: Run only on main branch
  if: github.ref == 'refs/heads/main' # github.ref: the full Git ref that triggered the workflow
  run: echo "Main branch deployment"

# The workflow was triggered by a pull_request event
- name: Run on PR
  if: github.event_name == 'pull_request'
  run: echo "PR validation"
```

- if: controls **whether a step runs**
- Conditions are written using **GitHub Actions expressions**
- Skipped steps are clearly marked in the Actions UI

| Common contexts used in if | Example | Meaning |
| ---  | ---  | ----  |
| github.ref | refs/heads/main | Branch |
| github.event_name | pull_request | Trigger type |
| github.actor | Username | Who triggered |
| github.repository | org/repo | Repo name |
| github.sha | Commit SHA | Commit ID |

Conditional execution in GitHub Actions allows steps or jobs to run only under specific conditions, such as branch, event type, or actor. Conditions are evaluated using workflow context expressions, enabling flexible CI/CD control without duplicating workflows.

#### Caching Dependencies
Saving downloaded packages between CI runs so installs are faster.  

Why it exists
- Avoid re-downloading the same wheels every run
- Reduce CI time
- Reduce network usage    

📌 Caching does NOT change what you test, it only speeds it up.

```yaml
- name: Cache Python dependencies
  uses: actions/cache@v3
  with:
    path: ~/.cache/pip
    key: ${{ runner.os }}-pip-${{ hashFiles('**/requirements.txt') }}
    restore-keys: |
      ${{ runner.os }}-pip-
```

almost always use Matrix Strategy (dependency variations) and Caching Dependencies together:  
```yaml
strategy:
  matrix:
    python-version: ['3.10', '3.11']
    deps: [min, latest]

- uses: actions/setup-python@v4
  with:
    python-version: ${{ matrix.python-version }}

- uses: actions/cache@v3
  with:
    path: ~/.cache/pip
    key: |
      pip-${{ matrix.python-version }}-${{ matrix.deps }}-${{ hashFiles('requirements*.txt') }}

- run: pip install -r requirements-${{ matrix.deps }}.txt
```



#### Fail fast control  

fail-fast applies **only to matrix jobs**. 
```yaml
jobs:
  test:
    runs-on: ubuntu-latest
    strategy:
      fail-fast: true # default
      matrix:
        python-version: ['3.10', '3.11']
```

By default, if any one matrix combination fails:
- All in-progress matrix jobs are cancelled
- Remaining ones are never started   

To get complete test coverage, one failure won’t cancel the others:  

```yaml
strategy:
  fail-fast: false
```

When to use **fail-fast: false**     

- Compatibility testing
- Dependency variations
- Cross-version / cross-OS testing
- CI signal quality > speed

You want to know:     
- Which versions fail
- How many fail
- Whether failures are isolated or systemic


When to keep **fail-fast: true**

- Expensive CI
- Smoke tests
- Deploy pipelines
- “One failure blocks everything” logic

Example:
- Security scans
- Production deploys


 



## Deployment Strategies

### 1. Blue-Green Deployment
Blue–green deployment maintains two identical production environments. The new version is deployed and tested in the inactive environment, and traffic is switched only after validation. This provides **zero (or near-zero) downtime and instant rollback** at the cost of running duplicate infrastructure.

<img src='./images/deployment_blue_green.jpg' width=500>

**Features**:  
- Maintain two identical production environments (Blue and Green)
- Deploy new version to inactive environment
- Switch traffic after validation
- **Advantage**: Instant rollback, zero downtime
- **Disadvantage**: Requires double infrastructure

**Steps**:
1. **Blue is live**
	  - Blue environment serves 100% of production traffic
	  - Green environment is idle or running an old version
2. **Deploy to Green**
    - Deploy the new version (v2) to Green
    - **No users are affected**
3. **Test Green**
    - Run:
      - Health checks
      - Smoke tests
      - Integration tests
    - Validate:
      - Startup
      - DB connections
      - External APIs
4. **Switch traffic**
    - Update the router / load balancer / DNS
    - Traffic moves from Blue → Green instantly
5. **Rollback (if needed)**
    - If something goes wrong:
      - Switch traffic **back to Blue**
      - No redeploy needed
      - Rollback is immediate

**Why it’s powerful**

- ✅ Zero downtime: No restarts on the live environment.
- ✅ Instant rollback: Rollback = traffic switch, not redeploy.
- ✅ Safe deployments: New version is fully validated before exposure.

**When to use Blue–Green**

- ✅ Production systems requiring **high availability**
- ✅ Systems needing **fast rollback**
- ✅ **APIs and backend services**
- ❌ Very cost-sensitive systems
- ❌ Apps with heavy state coupling

### 2. Rolling Deployment  

gradually replace old instances with new ones, keeping the service available the whole time without running two full environments.  

<img src='./images/deployment_rolling.jpg' width=500>

core idea: update production in small batches, so some instances run the old version while others run the new one until the rollout completes

**Features:**   
- Gradually replace instances with new version
- Update servers one-by-one or in batches
- **Advantage**: No extra infrastructure needed
- **Disadvantage**: Mixed versions during deployment

**Why rolling deployments are popular**
- ✅ No full duplicate environment: Lower cost than blue–green.
- ✅ Zero or near-zero downtime: Traffic always has healthy instances.
- ✅ Simple to implement

**Default strategy in**:
- Kubernetes
- AWS ECS
- Cloud Foundry

**When to use rolling**
- ✅ Stateless services
- ✅ Backward-compatible APIs
- ✅ Cost-sensitive systems
- ✅ High-frequency deployments
- ❌ Breaking API changes
- ❌ Large schema migrations
- ❌ Strong consistency requirements

#### Important trade-offs (this is where rolling can hurt)

- ⚠ **Mixed versions in production**      
    For a period, users hit both v1 and v2.     
    This can cause issues if:   
    - APIs change
    - DB schema is incompatible
    - Caches differ
- ⚠ **Rollback is slower**     
    To rollback:
    - Roll forward again to the old version
    - Takes time and affects traffic
- ⚠ **Capacity dip risk**     
    If not configured correctly, you may reduce capacity during rollout.


#### Health checks are critical

Rolling depends on:     
- Readiness probes
- Liveness probes

Traffic is sent **only to ready instances**.

#### Database compatibility rule (very important)

Rolling deployments require **backward-compatible changes**:

- ✅ Add columns
- ❌ Drop or rename columns
- ❌ Change column meaning

Use: Expand → migrate → contract


Rolling deployment updates instances incrementally, maintaining availability while avoiding duplicate environments. It’s cost-efficient but temporarily runs mixed versions and has slower rollback compared to blue–green.

### 3. Canary Deployment

release a new version to **a small subset of users first**, observe it, and **gradually increase traffic** only if everything looks healthy.

<img src='./images/deployment_canary.png' width=500>

**Features**:
- Deploy to small subset of users first
- Monitor metrics before full rollout
- **Advantage**: Risk mitigation, early feedback
- **Disadvantage**: Complex routing and monitoring

**Steps**:    
Assume v1 is live.

1. **Deploy canary**
    - Deploy v2 alongside v1
    - Route 5% of traffic to v2
2. **Observe metrics**   

    Monitor:
    - Error rate
    - Latency
    - CPU / memory
    - Business KPIs (checkout, signup)
3. **Increase traffic to v2 gradually**, if it is healthy
4. **Rollback instantly (if needed)**        

    - If metrics degrade:
      - Route traffic back to v1
      - Remove v2
    - Rollback affects only a small % of users.

**Why canary is powerful**

- ✅ Risk reduction: Only a small number of users are exposed to bugs.
- ✅ Real production validation: Uses real traffic, not just tests.
- ✅ Fast rollback: Traffic shift is quick and reversible.

**When to use canary**

- ✅ High-risk changes
- ✅ User-facing systems
- ✅ Large-scale distributed systems
- ✅ Performance-sensitive features
- ❌ Very small teams without monitoring
- ❌ Simple internal tools

Canary deployment gradually exposes a new version to a small percentage of users while monitoring key metrics. This minimizes blast radius, enables fast rollback, and provides real production validation at the cost of higher operational complexity.  

more: [Create a pipeline with canary deployments for Amazon ECS using AWS App Mesh](https://aws.amazon.com/blogs/containers/create-a-pipeline-with-canary-deployments-for-amazon-ecs-using-aws-app-mesh/)

### 4. Feature Flags

**Decouple deployment from release** by controlling features with configuration instead of code deploys. So **ship code safely**, turn features on/off instantly, and run experiments without redeploying.  

```text
Traditional flow:
Code change → Deploy → Feature live (risky)

With feature flags:
Code change → Deploy (hidden) → Enable safely → Roll back instantly
```

<img src='./images/deployment_feature_flag.avif' width=500>

**Features**:
- Deploy code with features disabled
- Enable features via flags for specific users/groups, environments, or percentages
- **Advantage**: Decouple deployment from release
- **Disadvantage**: Code complexity increases

Steps:
1. **Add a flag in code**
   ```python
    if feature_flags.is_enabled("new_checkout"):
        show_new_checkout()
    else:
        show_old_checkout()
   ```
2. **Deploy to production**
  - Code is deployed
  - Feature is off
  - Zero user impact
3. **Enable the flag**      
  Enable for:
  - Internal users
  - 1% of traffic
  - A specific region
4. **Monitor & adjust**
  - Increase rollout
  - Disable instantly if issues arise

**Types of feature flags**
| Type | Purpose |
|  |   |
| Release flags | Temporary rollout |
| Ops flags | Kill switches |
| Experiment flags | A/B testing |
| Permission flags | Role-based access |

**Common use cases**      

- ✅ Safe releases: Turn off instantly without redeploying.
- ✅ Gradual rollouts: Enable for 5% → 20% → 100%.
- ✅ A/B testing: Compare metrics between flag-on vs flag-off users.
- ✅ Kill switches: Disable broken features immediately.

**Feature flags + Canary (best practice)**      

They complement each other:     
- Canary → controls who gets v2    
- Feature flags → controls what is enabled      

Example:
- Canary deploy v2 to 10%
- Feature flag enables new search for only 2%

Comparision

| Strategy | Description | Rollback Speed | Risk Level | Traffic Split | Use Case |
|   ---|----|    -----|  ----  |  -----   |  -----|

| Feature Flags | Code deployed but features toggled on/off | Instant | Very Low | Configurable % | A/B testing, gradual feature rollout |

Agile (/'ædʒl/) is a methodology focused on **iterative development**, collaboration, and flexibility.

#### Agile vs Waterfall
**Waterfall**:   
- Linear, sequential process
- finish one phase before moving to the next.  
- Requirements → Design → Implementation → Testing → Deployment → Maintenance
- **Waterfall is better when**:
    - Requirements are very stable
    - Regulatory or contract-driven projects
    - Hardware, construction, defense
    - Changes are costly or dangerous
    - e.g. A medical device firmware with FDA approval requirements.
- **risk**: You find out late that you built the wrong thing.

**Agile**:
- Iterative, incremental process
- deliver small pieces continuously and adapt to change.
- Plan → Build → Test → Review → Iterate (every sprint)
- **Agile is better when**:
    - Requirements are uncertain or evolving
    - Need fast user feedback
    - Software products & SaaS
    - Competitive, fast-moving markets
    - e.g. A consumer web app or internal analytics platform.
- **risk**: You might move fast, but poor discipline can cause scope creep.

#### What is a sprint? 

A sprint is **a fixed time period** (usually 1-4 weeks, most commonly 2 weeks) during which a Scrum team works **to complete a set amount of work**.

**Key characteristics:**    

- **Time-boxed**: Fixed duration that doesn't change. Once you start a 2-week sprint, it stays 2 weeks
- **Goal-oriented**: Each sprint has a specific Sprint Goal that guides what the team builds
- **Potentially shippable increment**: By the end, the team should have working software that could theoretically be released
- **Consistent rhythm**: Sprints run back-to-back with no gaps, creating a predictable cadence

#### Story Points & Estimation
Story Points are **a unit of measure** used **to estimate the relative effort required to complete a user story**. They represent complexity, effort, and uncertainty, not time.  

***What story points measure***:   
- **Complexity**: How difficult is the work?
- **Effort**: How much work is involved?
- **Uncertainty**: How much is unknown?

**NOT hours or days** - they're relative, not absolute.  

<img src='./images/story_points_1.png' width=600>

***Common scales***:

- **Fibonacci Sequence**(most popular): 1, 2, 3, 5, 8, 13, 21
  - Represents relative complexity, not time
  - Smaller numbers = more precise estimates
  - Larger numbers = more uncertainty
  - Why Fibonacci: The gaps get bigger as numbers increase, reflecting growing uncertainty in larger estimates
- **T-shirt sizes** (alternative): XS, S, M, L, XL, XXL

<img src='./images/Story-point-estimation-cheat-sheet.png' width=600>

***Estimation techniques***:
1. **Planning Poker**

    - Most popular method
    - Each team member has cards with story point values
    - Product Owner reads user story
    - Team discusses briefly
    - Everyone simultaneously reveals their independent estimate
    - If estimates differ significantly, discuss and re-vote
    - Repeat until consensus
    
    <img src='./images/PlanningPoker.gif' width=600>

2. **T-shirt sizing**

    - Quick, less precise
    - Team assigns sizes, then converts to points later

3. **Relative estimation**

    - Compare new stories to already-estimated ones
    - "Is this bigger or smaller than story X?"

***Key concepts***:   
**Velocity**: Total story points completed per sprint

- Sprint 1: 23 points
- Sprint 2: 27 points
- Sprint 3: 25 points
- **Average velocity**: ~25 points/sprint

This helps predict how much work the team can commit to in future sprints.

***Why not estimate in hours?***    

- Developers work at different speeds
- Hours feel like commitments, points feel like estimates
- Points account for unknowns better
- Easier to estimate relative difficulty than absolute time

***Common pitfalls***:

- ❌ Comparing velocities between teams (**points are team-specific**)
- ❌ Using points for individual performance metrics
- ❌ Converting points back to hours ("1 point = 4 hours")
- ❌ Estimating stories over 13 points (break them down instead)

## Debugging Fundamentals

### What is Debugging?

**Definition**: The process of **finding and fixing bugs (defects)** in software code.

**Etymology**: The term "debugging" originated from Admiral Grace Hopper's team removing an actual moth from a computer in 1947.

#### Bug vs Defect vs Error

| Term | Definition | Example |
|------|------------|---------|
| **Error** | Human mistake in code | Typo, wrong logic |
| **Defect/Bug** | Flaw in code that causes incorrect behavior | Off-by-one error, null pointer |
| **Failure** | Observable incorrect behavior | App crashes, wrong output |

**Relationship**:
```text
Error (human mistake) 
    ↓
Defect (code flaw)
    ↓
Failure (observable problem)
```



### The Debugging Process

```text
1. Reproduce the bug
    ↓
2. Isolate the problem
    ↓
3. Identify the root cause
    ↓
4. Fix the issue
    ↓
5. Verify the fix
    ↓
6. Prevent regression (add tests)
```


### Types of Bugs

#### 1. Syntax Errors (Compile-time)
```python
# Missing colon
def calculate_total(items)
    return sum(items)

# Error: SyntaxError: invalid syntax
```
**Detection**: Caught by compiler/interpreter immediately



#### 2. Runtime Errors
```python
def divide(a, b):
    return a / b

result = divide(10, 0)
# Error: ZeroDivisionError: division by zero
```
**Detection**: Occurs during execution



#### 3. Logic Errors (Semantic Errors)
```python
def calculate_average(numbers):
    total = sum(numbers)
    return total / len(numbers) + 1  # Bug: Should not add 1
```
**Detection**: Code runs but produces wrong results



#### 4. Integration Bugs
```python
# Service A expects integer user_id
service_a.get_user(user_id=123)

# Service B sends string user_id
service_b.send_request(user_id="123")
# Bug: Type mismatch causes failure
```


#### 5. Concurrency Bugs (Race Conditions)
```python
# Two threads accessing shared resource
balance = 100

def withdraw(amount):
    global balance
    if balance >= amount:
        # Race condition here!
        balance -= amount
        
# Thread 1: withdraw(60)
# Thread 2: withdraw(60)
# Bug: Both might succeed, balance becomes -20
```


#### 6. Memory Leaks
```python
# Objects not properly cleaned up
cache = {}

def process_data(id, data):
    cache[id] = data  # Never removed!
    # Bug: Cache grows indefinitely
```



## Debugging vs Testing

### Key Differences

| Aspect | Testing | Debugging |
|--------|---------|-----------|
| **Goal** | Find if bugs exist | Find and fix bugs |
| **Who** | Testers, QA | Developers |
| **When** | Planned, systematic | Reactive, as needed |
| **Mindset** | Break the code | Fix the code |
| **Output** | Bug reports | Fixed code |
| **Process** | Repeatable, automated | Investigative, manual |


### The Testing-Debugging Cycle

```text
Write Code
    ↓
Run Tests ←─────────┐
    ↓               │
Test Fails?         │
    ↓ Yes           │
Debug & Fix         │
    ↓               │
Verify Fix ─────────┘
    ↓ No
Deploy
```


### How Testing Helps Debugging

**1. Narrows the scope**:
```python
# Without tests: Bug could be anywhere in 10,000 lines

# With unit tests: 
# ✅ test_calculate_total passed
# ❌ test_calculate_discount failed
# → Bug is in calculate_discount function
```

**2. Provides reproduction steps**:
```python
def test_discount_bug():
    """This test reproduces the bug"""
    result = calculate_discount(price=100, user_type="premium")
    assert result == 80  # Fails, returns 90
```

**3. Prevents regression**:
```python
# After fixing bug, test ensures it doesn't come back
def test_discount_premium_user():
    assert calculate_discount(100, "premium") == 80
```

### Profiling Tools (Performance Debugging)

#### 1. cProfile (Built-in)

- cProfile is Python’s **built-in performance profiler**.
- If pdb answers “what is my code doing?”, then cProfile answers “where is my code slow?” 🐢➡️🚀

cProfile measures:
- how many times each function is called
- how much time each function takes
- where your program spends its time

It’s:
- ✅ built into Python (no install)
- ✅ low overhead
- ✅ suitable for real programs (unlike profile, which is slower)

Key columns:   
- ncalls: Number of calls
- tottime: Time inside function (excluding subcalls)
- cumtime: Time including subcalls
- percall: Average time per call

Profile an entire script:  
```bash
python -m cProfile my_script.py
# Sort by cumulative time:
python -m cProfile -s cumtime my_script.py
```

✅ Use cProfile when:
- code is working but slow
- **Kafka consumer lag spikes**
- batch job takes too long
- API response time is high
- you need hard numbers, not guesses

❌ Don’t use it for:
- debugging logic errors → use pdb
- memory leaks → use tracemalloc
- async event loops → use async-aware profilers

````python
import cProfile

def slow_function():
    total = 0
    for i in range(1000000):
        total += i
    return total

# Profile the function
cProfile.run('slow_function()')

# Save results for later analysis: 
cProfile.run("slow()", "output.prof")
# Visualize it:
````

Visualize the output:  

```bash
pip install snakeviz
snakeviz output.prof
```

#### 2. line_profiler (Line-by-line)

- a line-by-line performance profiler for Python.
- If cProfile tells you which function is slow, line_profiler tells you **which exact line inside that function is slow** 🔍🐢 

It measures:
- execution time per line
- number of hits per line
- where time accumulates inside loops

It’s:
- ❌ not built-in
- ✅ extremely precise
- ⚠️ higher overhead (dev/debug only)

```bash
pip install line_profiler
```

⚠️ **Important limitations**:  
- ❌ Slows code significantly
- ❌ Not great for **async / multithreaded** code
- ❌ **Not for production**


use line_profiler when:
- data processing loops
- feature engineering pipelines
- Pandas-heavy transformations
- Kafka message processing logic
- algorithm optimization


````python
from line_profiler import LineProfiler

def slow_function():
    total = 0
    for i in range(10000000):
        total += i
    return total

lp = LineProfiler()
lp.add_function(slow_function)

lp.run('slow_function()')
lp.print_stats()
````

Key columns:
- Hits → how many times line ran
- Time → total time on that line
- % Time → where time is actually spent

🔥 That loop line is the bottleneck.

**Decorator-style (most common)**:

#### 3. memory_profiler

- A tool for **tracking memory usage over time** in Python.
- It tracks **RSS memory (real process memory)**, not Python object size.

  - pdb → what is my code doing?
  - cProfile → where is my code slow?
  - line_profiler → which line is slow?
  - memory_profiler answers: where is my code using too much memory?

It lets you:
- measure **memory usage per line**
- detect **memory leaks**
- see **memory growth over time**
- understand why a program eats RAM

✅ Use memory_profiler when:
- Pandas / NumPy code blows up RAM
- Kafka consumer slowly grows memory
- batch job OOMs
- caching logic leaks memory
- reading large files

❌ Don’t use it for:
- micro-optimizations
- CPU bottlenecks
- production (too slow)

```bash
pip install memory_profiler psutil
# psutil is optional but makes it more accurate & faster
```


**Line-by-line memory profiling (classic)**:  

```python
    from memory_profiler import profile

    @profile
    def build_list():
        data = []
        for i in range(1_000_000):
            data.append(i)
        return data

    build_list()
```
run with:  

```bash
    python -m memory_profiler my_script.py
```

Sample output:  
```text
    Line #   Mem usage   Increment   Occurrences   Line Contents
    ============================================================
        5   52.3 MiB    0.0 MiB     1             data = []
        6   95.8 MiB   +43.5 MiB    1000000       data.append(i)
        7   95.8 MiB    0.0 MiB     1             return data
```
Key columns:
- Mem usage → total memory at that line
- Increment → memory change caused by that line
- Occurrences → how many times line executed

🔥 That append line is the memory hog.


**Track memory over time (process-level)**

````python
from memory_profiler import memory_usage

def work():
    a = [i for i in range(10000000)]
    return a

mem = memory_usage(work, interval=0.1)
print(mem)
````

**memory_profiler vs tracemalloc**  

- memory_profiler tells you that memory is leaking;
- tracemalloc tells you where in Python it’s leaking.

| Tool| Tracks| 
| ----| -------| 
| memory_profiler| Process RSS memory| 
| tracemalloc (built-in)| Python object allocations| 

Use both when chasing serious leaks:  

```python
# memory_example.py
from memory_profiler import profile
import tracemalloc

@profile
def leaky_function():
    tracemalloc.start()

    data = []
    for i in range(100_000):
        data.append("x" * 1000)  # allocate strings

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics("lineno")

    print("\nTop Python allocations:")
    for stat in top_stats[:3]:
        print(stat)

    tracemalloc.stop()
    return data

leaky_function()
```

**What memory_profiler shows:**  
```text
Line #   Mem usage   Increment   Line Contents
=============================================
     8   55.1 MiB    0.0 MiB     data = []
     9  155.4 MiB  +100.3 MiB    data.append("x" * 1000)
```
✅ Tells you:   
- Memory jumped +100 MB   
- Problem is this line

❌ Doesn’t tell you what was allocated  

**What tracemalloc shows:**   
```text
Top Python allocations:
memory_example.py:9: size=95.4 MiB, count=100000, average=1000 B
```
✅ Tells you:
- Exact line allocating memory
- Object count
- Average object size

❌ Doesn’t tell you total process memory

**If memory grows but tracemalloc shows nothing** → suspect:
- NumPy
- Pandas
- C extensions
- native buffers

Minimal combo template: 
```python
from memory_profiler import profile
import tracemalloc

@profile
def target():
    tracemalloc.start()

    # suspicious code here

    snapshot = tracemalloc.take_snapshot()
    for stat in snapshot.statistics("lineno")[:5]:
        print(stat)

    tracemalloc.stop()

target()
```

**Typical profiling workflow (pro style)**

- 1️⃣ cProfile → find slow areas
- 2️⃣ line_profiler → optimize loops
- 3️⃣ memory_profiler → control RAM
- 4️⃣ tracemalloc → pinpoint leaks

### 2. Debugging Mockito Tests (Java)

#### Common Mockito Debugging Issues

**Issue 1: Mock Not Being Called**
```java
@Test
public void testUserService() {
    UserRepository mockRepo = mock(UserRepository.class);
    when(mockRepo.findById(1)).thenReturn(new User("Alice"));
    
    UserService service = new UserService(mockRepo);
    User user = service.getUser(1);
    
    // Debug: Was the mock actually called?
    verify(mockRepo).findById(1);  // This will fail if not called
}
```


**Issue 2: Wrong Argument Matchers**
```java
// ❌ This will fail
when(mockRepo.findById(any())).thenReturn(user);
service.getUser(1);
verify(mockRepo).findById(1);  // Fails! Expected any(), got 1

// ✅ Correct
when(mockRepo.findById(anyInt())).thenReturn(user);
// OR
when(mockRepo.findById(1)).thenReturn(user);
```


**Issue 3: Final or Static Methods**
```java
// ❌ Cannot mock final methods with standard Mockito
public final void process() { ... }

// ✅ Solutions:
// 1. Remove 'final' keyword
// 2. Use PowerMock
// 3. Use Mockito with inline mocking (Mockito 2+)
```



**Debugging Mockito Tests Checklist**:

- Verify mock behavior matches real usage
- Check argument matchers (any(), eq(), anyInt())
- Verify invocation counts (times(), never(), atLeast())
- Ensure methods aren't final or static
- Check if test is too brittle
- Consider integration test instead if mocking is too complex

### 3. Debugging Flaky Tests

**What are Flaky Tests?**
Tests that sometimes pass, sometimes fail (non-deterministic)

**Common Causes**:
1. **Race conditions**
2. **Timing issues**
3. **Random data**
4. **External dependencies**
5. **Test order dependency**


#### Debugging Race Conditions
```python
import threading

# Flaky test due to race condition
def test_concurrent_updates():
    counter = Counter()
    
    def increment():
        for _ in range(1000):
            counter.value += 1
    
    t1 = threading.Thread(target=increment)
    t2 = threading.Thread(target=increment)
    t1.start()
    t2.start()
    t1.join()
    t2.join()
    
    # Sometimes passes, sometimes fails!
    assert counter.value == 2000

# Fix: Add locking
class Counter:
    def __init__(self):
        self.value = 0
        self.lock = threading.Lock()
    
    def increment(self):
        with self.lock:
            self.value += 1
```



#### Debugging Timing Issues
```python
# Flaky test due to timing
def test_async_operation():
    result = start_async_task()
    time.sleep(0.1)  # ❌ Might not be enough!
    assert result.is_complete()

# Fix: Use polling with timeout
def test_async_operation_fixed():
    result = start_async_task()
    
    timeout = 5
    start = time.time()
    while not result.is_complete():
        if time.time() - start > timeout:
            pytest.fail("Timeout waiting for completion")
        time.sleep(0.1)
    
    assert result.is_complete()
```


#### Debugging Random Data Issues
```python
import random

```

### Root Cause Analysis

#### The 5 Whys Technique

**Problem**: User can't log in

```text
Why #1: Login fails with "Invalid credentials"
    ↓
Why #2: Because password hash doesn't match
    ↓
Why #3: Because hashing algorithm changed in deployment
    ↓
Why #4: Because migration script wasn't run
    ↓
Why #5: Because deployment process doesn't include migrations
    ↓
ROOT CAUSE: Missing migration step in deployment checklist
```

**Fix**: Add migration step to deployment process





### Reproducing Bugs Systematically

**Steps to Reproduce**:
1. **Identify minimum reproduction case**
2. **Isolate variables**
3. **Create automated test**
4. **Verify fix doesn't break test**

**Example**:
```python
# Bug report: "App crashes when processing orders"

# Crashes with order_id=12345, not with order_id=67890

def test_order_with_null_shipping_address():

    result = process_order(order)
    assert result is not None

# Step 4: Fix bug
def process_order(order):
    if order.shipping_address is None:
        raise ValueError("Shipping address required")
    # ... rest of processing
```




### Debugging by Elimination

**Process**:
1. List all possible causes
2. Systematically eliminate each
3. What remains is the bug

**Example**:
```text
Bug: API returns 500 error

Possible causes:
1. Database connection issue     → Test: Can connect ✓
2. Invalid request data          → Test: Valid JSON ✓
3. Missing authentication        → Test: Token present ✓
4. External API timeout          → Test: No external calls ✓
5. Null pointer in business logic → Test: Found null check missing! ✗

Root cause: Missing null check
```

### Intermittent Failures

**Causes**:
- Race conditions
- External dependencies
- Timing issues
- Resource contention

**Debug Steps**:
```python
# Add extensive logging
import logging
import time

def flaky_function():
    start = time.time()
    logger.debug(f"Start: {start}")
    
    try:
        result = external_call()
        elapsed = time.time() - start
        logger.debug(f"Success in {elapsed}s: {result}")
        return result
    except Exception as e:
        elapsed = time.time() - start
        logger.error(f"Failed after {elapsed}s: {e}")
        raise

# Run many times to catch pattern
for i in range(100):
    try:
        flaky_function()
    except:
        print(f"Failed on iteration {i}")
```

### Performance Issues

### Debugging Checklist

**When you encounter a bug**:

```text
□ Can you reproduce it?
□ What's the expected behavior?
□ What's the actual behavior?
□ What changed recently?
□ Does it happen consistently or intermittently?
□ What's the smallest code that reproduces it?
□ Have you checked logs?
□ Have you added print/log statements?
□ Have you stepped through with debugger?
□ Have you written a test that fails?
□ Have you searched for similar issues?
□ Have you asked for help (rubber duck)?
```



### Best Practices Summary

**Do**:
- ✅ Use debugger instead of print statements
- ✅ Add logging to production code
- ✅ Write tests that reproduce bugs
- ✅ Use version control (git bisect)
- ✅ Document your debugging process
- ✅ Ask for help when stuck

**Don't**:
- ❌ Make random changes hoping to fix it
- ❌ Debug without understanding the code
- ❌ Skip writing tests after fixing
- ❌ Leave debug code in production
- ❌ Give up too quickly
- ❌ Debug in production without backups
