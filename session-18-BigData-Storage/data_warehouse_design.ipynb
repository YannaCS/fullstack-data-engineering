## SCD
**Slowly Changing Dimension** types define how we handle changes in dimension data over time.   

SCD types are not just about handling changes, but about defining trade-offs between data accuracy, history preservation, and query simplicity.

它不只是“处理变化数据”，而是：如何在“是否保留历史”和“怎么保留历史”之间做权衡

| 类型| 本质策略| 是否保留历史| 
| ----| ----| ------------| 
| Type 1| 覆盖（overwrite）| ❌ 不保留| 
| Type 2| 追加版本（append row）| ✅ 完整保留| 
| Type 3| 增加列（add column）| ⚠️ 部分保留| 

👉 SCD 来自：
- 维度建模（Dimensional Modeling）
- 数据仓库设计（Kimball 方法论）

👉 也就是说：它解决的是“数据如何建模”，而不是“数据有多大”  

when to consider it: 
- In OLTP systems, we prioritize fast updates and current state, so SCD is unnecessary.
- In OLAP systems, we need historical tracking and point-in-time analysis, so SCD becomes essential.
  - 只有“做分析的数据库”才需要 SCD

如果 OLTP 也要保留历史怎么办: 
- In OLTP systems, historical data is usually handled via **audit tables, versioning, or event sourcing**, rather than Slowly Changing Dimension.
- 为什么不用 SCD？
    - SCD is optimized for analytical workloads, while OLTP systems prioritize write performance and simplicity.
