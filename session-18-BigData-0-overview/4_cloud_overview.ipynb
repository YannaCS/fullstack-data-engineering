## Quick Interview Cheat Sheet

> **面试常见问题：** "How would you migrate this AWS pipeline to GCP/Azure?"

**典型数据管道映射 (Typical Data Pipeline Mapping):**

```
AWS:    S3  →  Glue (ETL)  →  Redshift  →  QuickSight
GCP:    GCS →  Dataflow    →  BigQuery  →  Looker
Azure:  ADLS → Data Factory →  Synapse   →  Power BI
```

**流式管道映射 (Streaming Pipeline Mapping):**

```
AWS:    Kinesis  →  Kinesis Analytics  →  Redshift / S3
GCP:    Pub/Sub  →  Dataflow           →  BigQuery / GCS
Azure:  Event Hubs → Stream Analytics  →  Synapse / ADLS
```

**关键差异点 (Key Differentiators):**

| Point | AWS | GCP | Azure |
|-------|-----|-----|-------|
| DW Pricing Model | Per-node cluster | Per-query (serverless) | Dedicated + serverless |
| Serverless SQL | Athena (separate) | BigQuery (built-in) | Synapse Serverless (built-in) |
| Streaming | Kinesis (proprietary) | Pub/Sub (proprietary) | Event Hubs (Kafka-compatible) |
| ETL Paradigm | Glue (Spark-based) | Dataflow (Beam-based) | Data Factory (GUI + code) |
| Strongest BI | QuickSight (basic) | Looker (semantic layer) | Power BI (enterprise standard) |
